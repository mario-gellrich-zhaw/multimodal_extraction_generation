2. **Create:** let the OpenAI API create one picture per prompt.

## Setup

Select the **Python 3.11** kernel. The OpenAI API key comes from your `OPENAI_API_KEY` Codespaces secret. This notebook uses the **OpenAI API** (models `gpt-image-2` to create pictures and `gpt-6-luna` to rate them); other providers, e.g. Google Gemini or Anthropic Claude, have their own APIs with different code.

```python
# Python libraries.
import sys

from pydantic import BaseModel

# helper.py is in the main folder, one level up.
sys.path.append("..")
from helper import ask, create_pictures, show_images
```

## 1. Prompt from building blocks

Only **one** block changes between A and B, the rest stays the same – so you know what caused the difference.

```python
subject = "a small red wooden rowboat with two oars"
setting = "on a mountain lake"
light = "at dawn, soft light"
composition = "wide shot"
prompt_a = f"{subject}, {setting}, {light}, {composition}"

setting = "on a stormy ocean with high waves"  # change one block only
prompt_b = f"{subject}, {setting}, {light}, {composition}"

print("A:", prompt_a)
print("B:", prompt_b)
```

## 2. Create the pictures

`create_pictures()` lets the OpenAI API (model `gpt-image-2`, low quality) create one picture per name and prompt, about US$0.006 each, and saves them in `variants/`. An unchanged prompt is not created again, so you do not pay twice.

```python
paths = create_pictures(["example_a", "example_b"], [prompt_a, prompt_b], "variants")
show_images(paths, columns=2)
```

## 3. Rate one feature

The same yes/no question for every picture, answered in a fixed format.

```python
class Rating(BaseModel):
    matches: bool
    reason: str


PROMPT = "Does the image show stormy water, with high waves or a dark storm sky? Give a short reason."

for path in paths:
    print(path.name, ask(PROMPT, Rating, [path]))
```
